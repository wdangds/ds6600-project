```python
import numpy as np
import pandas as pd
import os
import requests
import json
import dotenv 
```

```python
dotenv.load_dotenv()  # Load environment variables from .env file
```

```python
congresskey = os.getenv("congresskey")
```

```python
root = 'https://api.congress.gov/v3'

endpoint = '/member'

params = {
    'format': 'json',
    'currentMember': 'true',
    'api_key': congresskey
}
```

```python
r = requests.get(root + endpoint, params=params)
```
