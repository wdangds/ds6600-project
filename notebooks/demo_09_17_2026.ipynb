record path -- somewhere in the JSON is a list where each item in one row of the dataframe you want to build. The record path is the sequence of keys to call to arrive at this list.

in this case the record path is just memebers

```python
params['limit'] = 250
r = requests.get(root + endpoint, params=params, headers=headers)
myjson = r.json()
congress_df = pd.json_normalize(myjson, record_path=['members'])
congress_df
```

```python
checkpoint = True
while checkpoint:
    # create upcomming request until no more data is available
    params['offset'] = len(congress_df)
    r = requests.get(root + endpoint, params=params, headers=headers)
    myjson = r.json()
    if len(myjson['members']) == 0:
        checkpoint = False
    else:
        new_df = pd.json_normalize(myjson, record_path=['members'])
        congress_df = pd.concat([congress_df, new_df], ignore_index=True)
```

```python
i = 0 
still_members = True
congress_json = []
while still_members:
    params['offset'] = i 
    r = requests.get(root + endpoint, params=params, headers=headers)
    myjson = r.json()
    if len(myjson['members']) == 0:
        still_members = False
    else:
        congress_json.extend(myjson['members'])
        i += len(myjson['members'])
congress_df = pd.json_normalize(congress_json)
congress_df
```

```python
congress_json = []
offset = 0

while True:
    params["offset"] = offset
    r = requests.get(root + endpoint, params=params, headers=headers)
    myjson = r.json()
    congress_json.extend(myjson["members"])
    offset += len(myjson["members"])
    if offset >= myjson["pagination"]["count"]:
        break
congress_df = pd.json_normalize(congress_json)
congress_df
```
