```python
import os 
import time 

for file in os.listdir('../data/raw'):
    start_time = time.time()
    # get time for all files loading , including csv, parquet, parquet.gzip, parquet.zstd, txt, parquet
    file_path = os.path.join('../data/raw', file)
    if file.endswith('.csv'):
        df = pd.read_csv(file_path)
        file_size = os.path.getsize(file_path)
    elif file.endswith('.parquet'):
        df = pd.read_parquet(file_path)
        file_size = os.path.getsize(file_path)
    elif file.endswith('.txt'):
        df = pd.read_csv(file_path, sep=r"\s*\|\s*,\s*\|\s*", engine='python', header = None)
        file_size = os.path.getsize(file_path)
    else:
        continue
    print(f"File: {file}, Size: {file_size} bytes, Load time: {time.time() - start_time:.4f} seconds")
```
