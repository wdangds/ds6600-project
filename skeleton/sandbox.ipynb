```python
import numpy as np 
import pandas as pd
import requests 

import plotly.express as px 
from dash import Dash, html, dcc
```

```python
url = 'https://api.congress.gov/v3/member'
param = {'api_key': 'wCZKBqTswprG2UM8ntlMC2x52KgRoZQfB3Q05aoo'}

r = requests.get(url, params = param)

r.json()
```

```python
memberdata = pd.json_normalize(r.json(), record_path = ['members'])
memberdata
```

```python
member_parties = memberdata.groupby('partyName').size().reset_index() # or directly reset_index(name='count')
member_parties = member_parties.rename(columns = {0: 'count'}) # or rename({0: 'count'}, axis = 1)
member_parties
```

```python
px.bar(member_parties, x='partyName', y='count')
```
